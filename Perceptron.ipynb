```python
#define the positive and negative values
positve_points=np.array([(-2,-3), (0,1), (-2,-1)])
negative_points=np.array([(-2,1), (0,-1), (2,-1)])
```

```python
x= np.vstack((positve_points, negative_points))
y= np.array([1]*len(positve_points)+[-1]*len(negative_points))
```

```python
#Perception algorithim
def perceptron(x,y,steps=10):
    theta=np.zeros(x.shape[1])
    theta0=0
    for T in range(steps):
        for i, xi in enumerate(x):
            if y[i]*(np.dot(xi,theta)+theta0)<=0:
                theta+=y[i]*xi
                theta0+=y[i]
                break
            #plot after each iteration
            plot_decision_boundary(theta, theta0, T+1)
            return theta, theta0
        
```
